```python
import pandas as pd

# Leitura do CSV
df = pd.read_csv(
    'ext-condicionamentos-nao-programados.csv',
    sep=',',
    quotechar="'",
)

# Converte colunas de data/hora (em milissegundos desde a época Unix)
df['data_hora_inicio'] = pd.to_datetime(df['data_de_início'], unit='ms')
df['data_hora_fim'] = pd.to_datetime(df['data_de_fim'], unit='ms')

# Extrai partes da data e hora de início
df['ano_inicio'] = df['data_hora_inicio'].dt.year.astype('Int32')
df['mes_inicio'] = df['data_hora_inicio'].dt.month.astype('Int32')
df['dia_inicio'] = df['data_hora_inicio'].dt.day.astype('Int32')
df['hora_inicio'] = df['data_hora_inicio'].dt.hour.astype('Int32')
df['minuto_inicio'] = df['data_hora_inicio'].dt.minute.astype('Int32')
df['segundo_inicio'] = df['data_hora_inicio'].dt.second.astype('Int32')

# Extrai partes da data e hora de fim
df['ano_fim'] = df['data_hora_fim'].dt.year.astype('Int32')
df['mes_fim'] = df['data_hora_fim'].dt.month.astype('Int32')
df['dia_fim'] = df['data_hora_fim'].dt.day.astype('Int32')
df['hora_fim'] = df['data_hora_fim'].dt.hour.astype('Int32')
df['minuto_fim'] = df['data_hora_fim'].dt.minute.astype('Int32')
df['segundo_fim'] = df['data_hora_fim'].dt.second.astype('Int32')

# Calcula o período de condicionamento
df['duracao'] = df['data_hora_fim'] - df['data_hora_inicio']

# Em horas, minutos e segundos (útil para análises)
df['duracao_horas'] = df['duracao'].dt.total_seconds() / 3600
df['duracao_minutos'] = df['duracao'].dt.total_seconds() / 60
df['duracao_segundos'] = df['duracao'].dt.total_seconds()

# Remove colunas desnecessárias
df.drop(columns=[
    'data_de_início',
    'data_de_fim',
    'motivo',
    'tipo_de_condicionamento',
    'arruamento',
    'troco',
    'entidade',
    'observacoes',
    'registo_data',
    'registo_hora',
    'ultima_modificacao_hora',
    'ultima_modificacao_data',
    'objectid',
    'esriGeometryPoint',
    'coordenada_x',
    'coordenada_y',
    'hora_de_inicio',
    'hora_de_fim',
    'data_hora_inicio',
    'data_hora_fim',
], inplace=True, errors='ignore')  # 'errors=ignore' evita erro se colunas faltarem

print(df.dtypes)
df.describe()
```
